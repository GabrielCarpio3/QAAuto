Imprimir por pantalla los números del 1 al 100 pero considerando lo siguiente: a) Si el número es divisible por 3 se debe imprimir "Fizz".
b) Si el número es divisible por 5 se debe imprimir "Buzz".
c) Si el número es divisible por 3 y por 5 se debe imprimir "FizzBuzz"

```python
def fizzBuzz():
  for i in range(1,101):
    if i % 3 == 0 and i % 5 == 0:
      print("FizzBuzz")
    elif i % 3 == 0:
      print("Fizz")
    elif i % 5 == 0:
      print("Buzz")
    else:
      print(i)

fizzBuzz()
```

```python
def fizzBuzz2(numero):
  if numero % 3 == 0 and numero % 5 == 0:
    return "FizzBuzz"
  elif numero % 3 == 0:
    return "Fizz"
  elif numero % 5 == 0:
    return "Buzz"
  else:
    return numero

def getLista(numero):
  lista = []
  for i in range(1,numero + 1):
    lista.append(fizzBuzz2(i))
  return lista

```

```python
import unittest

class fizzBuzzTest(unittest.TestCase):
  def testFizz(self):
    self.assertEqual("Fizz",fizzBuzz2(6))
  def testBuzz(self):
    self.assertEqual("Buzz",fizzBuzz2(10))
  def testFizzBuzz(self):
    self.assertEqual("FizzBuzz",fizzBuzz2(15))
  def testListaHasta15(self):
    esperado = [1,2,"Fizz",4,"Buzz","Fizz",7,8,"Fizz","Buzz",11,"Fizz",13,14,"FizzBuzz"]
    self.assertEqual(esperado,getLista(15))

unittest.main(defaultTest="fizzBuzzTest", argv=[""], verbosity=2, exit=False)
```

Los números expansivos se definen de la siguiente forma:

El primer número expansivo es el 1
Dado un número expansivo x, para calcular el siguiente se ha de sustituir cada grupo de n cifras consecutivas iguales que aparezcan en x (por ejemplo, 3333) por número n seguido de la cifra que se repite (en el ejemplo 43)

Así, los primeros números expansivos son:

1 -> "un uno": 11

11 -> "dos unos": 21

21 -> "un dos, un uno": 1211

1211 -> "un uno, un dos, dos uno": 111221

Se pide programar:

una función "expand" que, dado un arreglo con las cifras de un número expansivo genere el siguiente
una función "list2num" que, dado un arreglo con las cifras de un número expansivo, devuelva el número correspondiente

```python
import numpy as np

def expand(cifras: np.array):
    resultado = []
    i = 0
    n = len(cifras)
    while i < n:
      conteo = 1
      while i + 1 < n and cifras[i] == cifras[i + 1]:
        conteo += 1
        i += 1
      resultado.append(conteo)
      resultado.append(cifras[i])
      i += 1
    return resultado

def list2num(cifras) -> int:
    numero_str = "".join(str(c) for c in cifras)
    return int(numero_str)


entrada = [1, 2, 1, 1]
siguiente = expand(entrada)
print("Resultado expand :", siguiente)

numero = list2num(siguiente)
print("Resultado list2num (Entero):", numero)
```

```python
import unittest

class numerosExpasivosTest(unittest.TestCase):
  def test_expand_secuencia_estandar(self):
        """1. Verifica la expansión de [1, 2, 1, 1] -> [1, 1, 1, 2, 2, 1]"""
        entrada = np.array([1, 2, 1, 1])
        esperado = np.array([1, 1, 1, 2, 2, 1])
        np.testing.assert_array_equal(expand(entrada), esperado)

  def test_expand_caso_inicial(self):
        """2. Verifica la primera expansión [1] -> [1, 1]"""
        entrada = np.array([1])
        esperado = np.array([1, 1])
        np.testing.assert_array_equal(expand(entrada), esperado)

  def test_list2num_conversion_entero(self):
        """3. Verifica que la conversión a número entero sea correcta"""
        entrada = np.array([1, 1, 1, 2, 2, 1])
        self.assertEqual(list2num(entrada), 111221)

unittest.main(defaultTest="numerosExpasivosTest", argv=[""], verbosity=2, exit=False)
```

```python
%%file p_expansivos.py
import pytest
import numpy as np

# --- 1. Funciones a probar ---
def expand(cifras: np.array):
    resultado = []
    i = 0
    n = len(cifras)
    while i < n:
        conteo = 1
        while i + 1 < n and cifras[i] == cifras[i + 1]:
            conteo += 1
            i += 1
        resultado.append(conteo)
        resultado.append(cifras[i])
        i += 1
    return np.array(resultado)

def list2num(cifras: np.array) -> int:
    numero_str = "".join(cifras.astype(str))
    return int(numero_str)


# --- 2. Pruebas Unitarias con Pytest ---
def test_expand_secuencia():
    """Test 1: Verifica la expansión de [1, 2, 1, 1] -> [1, 1, 1, 2, 2, 1]"""
    actual = expand(np.array([1, 2, 1, 1]))
    expected = np.array([1, 1, 1, 2, 2, 1])
    np.testing.assert_array_equal(actual, expected)

def test_expand_caso_base():
    """Test 2: Verifica la expansión inicial de [1] -> [1, 1]"""
    actual = expand(np.array([1]))
    expected = np.array([1, 1])
    np.testing.assert_array_equal(actual, expected)

def test_list2num_conversion():
    """Test 3: Verifica la conversión de ndarray a número entero"""
    actual = list2num(np.array([1, 1, 1, 2, 2, 1]))
    expected = 111221
    assert actual == expected
```
