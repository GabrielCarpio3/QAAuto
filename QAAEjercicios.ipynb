# FizzBuzz



```python
def fizzbuzz(numero):
  if numero % 3 == 0 and numero % 5 == 0:
    return "FizzBuzz"
  elif numero % 3 == 0:
    return "Fizz"
  elif numero % 5 == 0:
    return "Buzz"
  else:
    return numero

def get_lista(numero):
  lista = []
  for i in range(1,numero + 1):
    lista.append(fizzbuzz(i))
  return lista

print(get_lista(100))
```

```python
import unittest

class fizzbuzz_test(unittest.TestCase):
  def test_fizz(self):
    self.assertEqual("Fizz",fizzbuzz(6))
  def test_buzz(self):
    self.assertEqual("Buzz",fizzbuzz(10))
  def test_fizzbuzz(self):
    self.assertEqual("FizzBuzz",fizzbuzz(15))
  def test_lista_hasta_5(self):
    esperado = [1,2,"Fizz",4,"Buzz"]
    self.assertEqual(esperado,get_lista(5))

unittest.main(defaultTest="fizzBuzzTest", argv=[""], verbosity=2, exit=False)
```

```python
# pytest ui/test_coffecart.py -v
import pytest
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support.wait import WebDriverWait

@pytest.mark.usefixtures('class_setup')
class Test_coffe:

  @pytest.fixture
  def class_setup(self):
    self.driver = webdriver.Chrome()
    self.driver.maximize_window()
    self.driver.get('https://coffee-cart.app/')

  def test_carrito_inicial_vacio(self):
   boton_carrito = self.driver.find_element(By.XPATH, "//button[@data-test='checkout']")

   assert '$0.00' in boton_carrito.text
   self.driver.quit()


  def test_agregar_un_producto(self):
    primer_cafe = WebDriverWait(self.driver, 10).until(EC.element_to_be_clickable((By.XPATH, "//div[@data-cy='Espresso']")))
    primer_cafe.click()

    boton_carrito = self.driver.find_element(By.XPATH, "//button[@data-test='checkout']")

    assert '$0.00' not in boton_carrito.text
    self.driver.quit()


  def test_agregar_varios_productos_en_cart(self):
    # Espresso ($10.00)
    espresso = WebDriverWait(self.driver, 10).until(EC.element_to_be_clickable((By.XPATH, "//div[@data-cy='Espresso']")))
    espresso.click()

    # Cappuccino ($19.00)
    cappuccino = WebDriverWait(self.driver, 10).until(EC.element_to_be_clickable((By.XPATH, "//div[@data-cy='Cappuccino']")))
    cappuccino.click()

    # Click a cart
    boton_cart_superior = WebDriverWait(self.driver, 10).until(EC.element_to_be_clickable((By.XPATH, "//a[@aria-label='Cart page']")))
    boton_cart_superior.click()

    # Precio total en cart ($10.00) + ($19.00) = ($29.00)
    boton_pago_carrito = WebDriverWait(self.driver, 10).until(EC.visibility_of_element_located((By.XPATH, "//button[@data-test='checkout']")))
    assert '$29.00' in boton_pago_carrito.text
    self.driver.quit()
```
